```python
from kcatbench import plotting
from kcatbench.util import DATA_DIR, RESULT_DIR
import pandas as pd
import ast
```

```python
result_dir = DATA_DIR / "results"

kcat_data = pd.read_csv(result_dir / "kcat_data_complete.csv")
```

```python
def string_to_list(text):
    try:
        if isinstance(text, list) or pd.isna(text):
            return text
        return ast.literal_eval(text)
    except (ValueError, SyntaxError):
        return None

kcat_data['mmkcat_kcat'] = kcat_data['mmkcat_kcat'].apply(string_to_list)
kcat_data['prod_smiles'] = kcat_data['prod_smiles'].apply(string_to_list)
```

```python
display_names = {
    'dlkcat': 'DLKcat', 
    'catapro': 'CataPro', 
    'turnup': 'TurNuP', 
    'unikp': 'UniKP', 
    'catpred': 'CatPred', 
    'mmkcat': 'MMKcat'
}

plotting.plot_model_intersection_sets(
    kcat_data,
    display_names,
    subset_type='best',
    save=True,
    show=False
)
```

```python
plotting.plot_model_intersection_sets(
    kcat_data,
    display_names,
    subset_type='worst',
    save=True,
    show=False
)
```

```python
models = [('dlkcat_kcat', 'DLKcat'), ('catapro_kcat', 'CataPro'), ('turnup_kcat', 'TurNuP'), ('unikp_kcat', 'UniKP'), ('catpred_kcat', 'CatPred'), ('mmkcat_kcat', 'MMKcat')]

for i, model1 in enumerate(models):
    plotting.plot_model_comparison(kcat_data, 'experimental_kcat', model1[0], 'Experimental', model1[1], log_scale=True, save=True, gridsize=70, show=False)
    if i < (len(models) - 1):
        for j in range(i + 1, len(models)):
            model2 = models[j]
            plotting.plot_model_comparison(kcat_data, model1[0], model2[0], model1[1], model2[1], log_scale=True, save=True, gridsize=70, show=False)
```

```python
import img2pdf

plot_dir = RESULT_DIR / "plots" / "comparison_plots"

files = sorted([str(f) for f in plot_dir.glob("*.png")])

with open("all_plots.pdf", "wb") as f:
    f.write(img2pdf.convert(files))

print(f"Created PDF with {len(files)} plots.")
```
